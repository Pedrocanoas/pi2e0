# 🚀 Training, Validation, and Prediction - YOLO11

```python
# === Imports ===
!pip install ultralytics -q  # Install the Ultralytics YOLO package silently

import os
import time
from datetime import datetime
from ultralytics import YOLO
import ultralytics

# === Configuration ===

# Path to the pretrained model used as the starting point (YOLOv11m in this case)
MODEL_PATH = 'yolo26l.pt'

# Path to the dataset configuration file (defines train/val/test splits and class names)
DATA_YAML = '/kaggle/input/glass-defect/config.yaml'

# Base dataset directory (same level as this notebook)
dataset_dir = '/kaggle/input/glass-defect/dataset/dataset'

# Training hyperparameters
EPOCHS = 500           # Maximum number of training epochs
PATIENCE = 250         # Stop training if no improvement after this many epochs
CONF_TRAIN = 0.4       # Minimum confidence threshold during training
CONF_PREDICT = 0.6     # Minimum confidence threshold for predictions
SINGLE_CLASS = True    # Set to True if training a single-class detection model

# Directory paths
IMAGE_TO_PREDICT_DIR = os.path.join(dataset_dir, 'images', 'test')  # Test image folder
RESULT_DIR = os.path.join('/kaggle/working/runs', 'results')        # Output folder for results
```

```python
# === Environment Check ===
print("🔍 Checking YOLO environment...")
ultralytics.checks()  # Verify that all YOLO dependencies and configurations are properly installed

print(f"📦 loading moodel: {MODEL_PATH}")
model = YOLO(MODEL_PATH)
```

```python
# --- Training ---
print("🚀 Starting model training...")

train_results = model.train(
    data=DATA_YAML,          # Path to dataset YAML file
    epochs=EPOCHS,           # Total number of training epochs
    patience=PATIENCE,       # Early stopping patience
    val=True,                # Enable validation during training
    batch=16,                # Batch size
    single_cls=SINGLE_CLASS, # Single-class training (defects)
    lr0=0.002,               # Initial learning rate
    device='cuda',           # Use GPU (CUDA)
)
```

```python
# --- Validation ---
print("📈 Evaluating model on validation set...")

val_results = model.val(
    conf=CONF_TRAIN,   # Minimum confidence threshold to consider a detection valid
    save_json=True     # Save metrics in COCO JSON format
)
```

```python
# --- Prediction ---
print(f"🖼️ Searching for images in: {IMAGE_TO_PREDICT_DIR}")

# Filter valid image extensions
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
image_paths = [
    os.path.join(IMAGE_TO_PREDICT_DIR, f)
    for f in os.listdir(IMAGE_TO_PREDICT_DIR)
    if f.lower().endswith(image_extensions)
]

if not image_paths:
    print(f"❌ No images found in {IMAGE_TO_PREDICT_DIR}")
else:
    print(f"🔍 {len(image_paths)} images found for prediction.")
    pred_results = model.predict(source=image_paths, conf=CONF_PREDICT)
```

```python
# --- Save Results ---
os.makedirs(RESULT_DIR, exist_ok=True)

for i, result in enumerate(pred_results):
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"result_{i+1}_{timestamp}.jpg"
    result_path = os.path.join(RESULT_DIR, filename)
    result.save(filename=result_path)
    print(f"💾 Result saved at: {result_path}")

print("✅ Execution completed successfully!")
```

```python
# --- Trained Weights Path ---
# IMPORTANT: this is the model that was actually fine-tuned on glass defects.
# Any downstream inference (real-time script included) must load THIS file,
# not the original MODEL_PATH ('yolo26l.pt'), which is just the generic
# pretrained starting point used before training.
BEST_WEIGHTS = os.path.join(train_results.save_dir, "weights", "best.pt")
print(f"🏋️ Trained weights saved at: {BEST_WEIGHTS}")
print("⬇️ Download this file from the Kaggle Output panel and use it as")
print("   MODEL_PATH in the real-time inspection script (realtime_inspection.py).")
```

# 🎯 Real-Time Defect Detection from Camera Stream

Kaggle notebooks don't have access to a live inspection camera, so the
real-time part of this project doesn't live here anymore. After training,
download `BEST_WEIGHTS` (printed above) and use it with the standalone
`realtime_inspection.py` script (in the project root), which runs on the PC
connected to the camera on the production line.

That script keeps the same multi-pass confirmation logic used before (a
defect is only reported once the same tracked box has been detected across
several consecutive frames as the glass crosses the camera's field of view),
but is built for actual real-time use:
- decoupled capture/inference/display threads, so a slow render never makes
  the camera feed lag behind;
- `cv2.imshow` instead of Matplotlib (a notebook render can take
  100–500ms/frame; a native window takes ~1–5ms);
- explicit device/image-size settings that are easy to tune for CPU-only
  hardware.
